```python
from datetime import datetime
def df_to_date(df):
    dates = [datetime(year=2020, month=row.Month, day=row.Day) for row in df.itertuples(index=False)]
    return dates
```

```python
import matplotlib.pyplot as plt
```

```python
y_range = range(1,9)

cols = ['Avg_Temperature', 'Radiation', 'Rain_Amount', 'Wind_Speed', 'Wind_Direction']
plt.figure(figsize=(12, 12))
for j, c in enumerate(cols):
    for i in y_range:
        sub_df = graph_df[graph_df['Year']==i]
        dates = df_to_date(sub_df)
        plt.subplot(len(cols),1, j+1)
        plt.plot(dates, sub_df[c], label=f'Year {i}',linewidth=1)
        if j==0:
            plt.legend()
plt.show()
```
